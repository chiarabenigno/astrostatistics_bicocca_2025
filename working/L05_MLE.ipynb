```python
import numpy as np
from tqdm.notebook import tqdm
import pylab as plt
from matplotlib import pyplot as plt
import math
from scipy.stats import norm
```

```python
mu, sigma = 1, 0.2
data = np.random.normal(1, 0.2, 1000)
plt.hist(data, bins = 100, density=True, alpha = 0.5, color = 'g', edgecolor = 'blue')
x = np.linspace(min(data), max(data), 1000)
pdf = (1/(sigma * np.sqrt(2 * np.pi)) * np.exp(- (x-media) ** 2 / (2 * sigma ** 2)))
plt.plot(x, pdf, 'r-', lw=2)

media = np.mean(data)
```

```python
#Possible mu from 0 to 2
mu_proposed = np.linspace(0, 2, 1000)
#Compute  lnL of each possible mu
lnL_scan = []
for mu in mu_proposed:
    dist = norm(loc=mu)
    #log probability density
    lnL_temp = dist.logpdf(data)
    #sum over log pdf of all data points
    lnL_temp = np.sum(lnL_temp)
    lnL_scan.append(lnL_temp)

#Convert to numpy array
lnL_scan = np.array(lnL_scan)

#mu_proposed that maximizes the lnL.
plt.plot(mu_proposed, lnL_scan)

find_max = mu_proposed[np.argmax(lnL_scan)]
print(find_max)
plt.axvline(find_max,c='red')

plt.axvline(media,c='black',ls='dotted')
```
